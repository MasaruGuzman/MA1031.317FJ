# Correlación
## Cargar librerías

```python
# librería para base de datoss
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.formula.api as smf
```

## Cargar la base de datos

```python
datos = pd.read_excel("/content/drive/MyDrive/MA1031.317FJ/datos_ventas_limpios(1).xlsx")
datos
```

## Revisar correlación entre variables

```python
datos.corr()
```

## Diagrama de dispersión


```python
plt.scatter(datos['Television'], datos['Ventas'])
plt.xlabel("Television")
plt.ylabel("Ventas")
plt.title("Diagrama de dispersion")
plt.show()
```

## Crear modelo de regresión

```python
modelo = smf.ols(formula = 'Ventas~Television', data = datos).fit()
print(modelo.summary())
```

## Ecuación de regresión (aprox)


```python
y_aprox = 7.0326 + 0.0475*datos['Television']
print(y_aprox)
```

## Errores

```python
errores = datos['Ventas'] - y_aprox
print(errores)
```

## Gráfica de los errores

```python
plt.scatter(range(200), errores)
plt.axhline(y = 0, color = "r")
plt.xlabel("Numero del error")
plt.ylabel("Errores")
plt.title("Gráfica de los errores")
plt.show()
```

Histograma de los errores


```python
plt.hist(errores)
plt.xlabel("Errores")
plt.ylabel("Frecuencia")
plt.title("Histograma de los errores")
plt.show()
```

## Comparación entre datos reales y aproximaciones

```python
plt.scatter(datos['Ventas'], y_aprox)
plt.plot(datos['Ventas'], datos['Ventas'], color = 'red')
plt.xlabel("Ventas reales")
plt.ylabel("Ventas aproximadas")
plt.
plt.title("Comparación entre datos reales y aproximaciones")
plt.show()
```
